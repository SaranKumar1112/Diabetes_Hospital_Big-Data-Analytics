PREDICTIVE ANALYSIS

```python
# Libraries import
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
```

```python
data_f = pd.read_csv('../data/updated cleaned_diabetic_data_with_Median.csv')
```

```python
X = data_f[['time_in_hospital', 'num_lab_procedures', 'num_medications', 'number_diagnoses',
            'discharge_disposition_id', 'number_outpatient', 'age', 'number_emergency', 'num_procedures']]
y = data_f['readmitted']

# Explicit mapping: NO=0, >30=1 (readmitted late), <30=2 (readmitted early/urgent)
y_encoded = y.map({'NO': 0, '>30': 1, '<30': 2})
```

```python
# Categorical/numerical columns 
categorical_features = ['discharge_disposition_id', 'age']
numerical_features = ['time_in_hospital', 'num_lab_procedures', 'num_medications',
                      'number_diagnoses', 'number_outpatient', 'number_emergency', 'num_procedures']
```

```python
# Data Preprocessing (num =numerical and cat=categorical features)
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(drop='first'), categorical_features)
    ])

# preprocessing applied
X_preprocessed = preprocessor.fit_transform(X)
```

```python
# Data Split
X_train, X_test, y_train, y_test = train_test_split(X_preprocessed, y_encoded, test_size=0.2, random_state=42)
```

```python
param_grid = {
    'C': [0.001, 0.01, 0.1, 1, 10],
    'solver': ['liblinear', 'saga']
}

# class_weight='balanced' addresses the severe class imbalance (<30 has ~11% of records)
grid_search = GridSearchCV(
    LogisticRegression(max_iter=5000, class_weight='balanced'),
    param_grid, cv=5,
    n_jobs=-1, scoring='accuracy'
)
grid_search.fit(X_train, y_train)
```

```python
# Get Best Parameters
best_params = grid_search.best_params_
cross_val_score = grid_search.best_score_
```

```python

# best param results
print(f'Best Parameters: {best_params}')
print(f'Cross Validation Score: {cross_val_score:.4f}')

# Prediction test set
y_pred = grid_search.best_estimator_.predict(X_test)

# metric calc
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=['NO', '>30', '<30'])
conf_matrix = confusion_matrix(y_test, y_pred)

# display results
print(f'Accuracy: {accuracy:.4f}')
print('Classification Report:')
print(report)
```

```python
# confusion matrix
plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['NO', '>30', '<30'], yticklabels=['NO', '>30', '<30'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```
